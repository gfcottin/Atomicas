# Cálculo de la vida media del muón (con relatividad)

```python
import math

C_M_S = 3.0e8

def distancia_recorrida(h1_m, h2_m):
    return abs(h1_m - h2_m)

def tiempo(distancia_m, vel_m_s):
    return distancia_m / vel_m_s

def segundos_a_microsegundos(t_s):
    return t_s * 1e6

def gamma_factor(vel_m_s, c_m_s=C_M_S):
    beta2 = (vel_m_s*vel_m_s)/(c_m_s*c_m_s)
    return 1.0/math.sqrt(1.0-beta2)

def tau_propia_desde_cuentas(t_s, gamma, N1, N2):
    return -(t_s/gamma)*(1/math.log(N1/N2))
```

```python
h1_m = 1850.0
h2_m = 547.0
N1   = 9.1
N2   = 2.8
v    = 0.9 * C_M_S

delta_h = distancia_recorrida(h1_m, h2_m)
t_vuelo = tiempo(delta_h, v)
t_vuelo_ns = segundos_a_microsegundos(t_vuelo)
gamma = gamma_factor(v)

tau0_s = tau_propia_desde_cuentas(t_vuelo, gamma, N1, N2)
tau0_ns = segundos_a_microsegundos(tau0_s)

print("Tiempo de vuelo (μs):", t_vuelo_ns)
print("Factor γ:", gamma)
print("Vida propia τ0 (μs):", tau0_ns)
```

## Comparación con PDG

```python
tau0_pdg_ns = 2.197
diff_pct = 100.0*(tau0_ns - tau0_pdg_ns)/tau0_pdg_ns

print("Estimación τ0 (ns):", tau0_ns)
print("Valor PDG τ0 (ns):", tau0_pdg_ns)
print("Diferencia relativa (%):", diff_pct)
```
