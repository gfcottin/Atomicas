
# Graficador de Contador de Muones

Este cuaderno lee uno o más archivos de texto plano donde cada línea tiene dos valores separados por coma:

```
<id_muon>,<tiempo_en_segundos>
```

Extrae:
- el **número secuencial de muones** (o cuenta acumulada), y
- el **tiempo de detección** (segundos),

y luego produce:
1. Gráficas individuales (tiempo vs. cuenta) para cada ubicación/conjunto de datos, y  
2. Una figura comparativa que superpone todos los conjuntos de datos.

> **Personaliza las rutas de los archivos en la sección `Principal`.**


## Importaciones

````python

import matplotlib.pyplot as plt
import numpy as np
from pathlib import Path
from typing import List, Tuple
````

## Funciones auxiliares (lectura y extracción de datos)

````python

def numero_muon(linea: str) -> int:
    """Devuelve el número secuencial del muón (primer campo CSV)."""
    partes = linea.strip().split(",")
    return int(partes[0])


def tiempo_muon(linea: str) -> int:
    """Devuelve el tiempo de detección del muón en segundos (segundo campo CSV)."""
    partes = linea.strip().split(",")
    return int(partes[1])


def extraer_datos(lineas: List[str]):
    """
    Dadas líneas tipo 'N,T', devuelve dos listas:
      - n_muon: números secuenciales de muones (int)
      - t_muon: tiempos en segundos (int)
    Ignora líneas vacías o malformadas.
    """
    n_muon = []
    t_muon = []
    for linea in lineas:
        linea = linea.strip()
        if not linea or "," not in linea:
            continue
        try:
            n_muon.append(numero_muon(linea))
            t_muon.append(tiempo_muon(linea))
        except Exception:
            # Ignora líneas malformadas silenciosamente
            continue
    return n_muon, t_muon
````

## Funciones de graficado

````python

def graficar_cuenta_muones(x_tiempo, y_cuenta, color, etiqueta):
    """Grafica una curva tiempo vs número de muones."""
    plt.figure(figsize=(9, 4.5))
    plt.plot(x_tiempo, y_cuenta, color=color, label=etiqueta)
    plt.xlabel("Tiempo (s)")
    plt.ylabel("Cuenta acumulada de muones")
    plt.title("Número de muones detectados en función del tiempo")
    plt.legend(loc="upper left")
    plt.tight_layout()
    plt.show()


def graficar_todos(
    x0, y0, color0,
    x1, y1, color1,
    x2, y2, color2,
    etiquetas=("Ubicación A", "Ubicación B", "Ubicación C"),
    limite_y=None
):
    """Superpone tres curvas tiempo vs cuenta en una sola figura para comparar."""
    plt.figure(figsize=(10, 5))
    plt.plot(x0, y0, color=color0, label=etiquetas[0])
    plt.plot(x1, y1, color=color1, label=etiquetas[1])
    plt.plot(x2, y2, color=color2, label=etiquetas[2])
    plt.xlabel("Tiempo (s)")
    plt.ylabel("Cuenta acumulada de muones")
    plt.title("Comparación de número de muones detectados en función del tiempo")
    if limite_y is not None:
        plt.ylim(0, limite_y)
    plt.legend(loc="upper right")
    plt.tight_layout()
    plt.show()
````
