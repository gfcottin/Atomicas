# Muon Lifetime Calculation (with relativity)

```python
import math

C_M_S = 3.0e8

def traveled_distance(h1_m, h2_m):
    return h2_m - h1_m

def time_of_flight(distance_m, vel_m_s):
    return distance_m / vel_m_s

def seconds_to_microseconds(t_s):
    return t_s * 1e6

def lorentz_gamma(vel_m_s, c_m_s=C_M_S):
    beta2 = (vel_m_s*vel_m_s)/(c_m_s*c_m_s)
    return 1.0/math.sqrt(1.0-beta2)

def proper_tau_from_counts(t_s, gamma, N1, N2):
    return -(t_s/gamma)/math.log(N1/N2)
```

```python
h1_m = 2793.0
h2_m = 547.0
N1   = 15.7
N2   = 2.8
v    = 0.9 * C_M_S

delta_h = traveled_distance(h1_m, h2_m)
t_flight = time_of_flight(delta_h, v)
t_flight_ns = seconds_to_microseconds(t_flight)
gamma = lorentz_gamma(v)

tau0_s = proper_tau_from_counts(t_flight, gamma, N1, N2)
tau0_ns = seconds_to_microseconds(tau0_s)

print("Flight time (μs):", t_flight_ns)
print("Lorentz factor γ:", gamma)
print("Proper lifetime τ0 (μs):", tau0_ns)
```

## Comparison with PDG

```python
tau0_pdg_ns = 2.197
diff_pct = 100.0*(tau0_ns - tau0_pdg_ns)/tau0_pdg_ns

print("Estimated τ0 (μs):", tau0_ns)
print("PDG τ0 (μs):", tau0_pdg_ns)
print("Relative difference (%):", diff_pct)
```
